```python
data_folder = "../Camera_Calibrations/Ximea_Camera"
gain_map = IO.read_tiff(os.path.join(data_folder, "gain.tif"))
offset_map = IO.read_tiff(os.path.join(data_folder, "offset.tif"))
variance = IO.read_tiff(os.path.join(data_folder, "variance.tif"))
read_noise = IO.read_tiff(os.path.join(data_folder, "readnoise.tif"))
rqe = IO.read_tiff(os.path.join(data_folder, "rqe.tif"))
R, G, B, wavelength = S_F.getpixelefficiency()

pixel_QYs = np.vstack([B, G, R])
camera_parameters = {}
camera_parameters["pixel_QYs"] = pixel_QYs
camera_parameters["pixel_order"] = ["B", "G", "R"]
camera_parameters["pixel_order_indices"] = [0, 1, 2]
```

```python
import types

smoothing_function = types.SimpleNamespace()
smoothing_function.args = {"sigma": 1.5}
smoothing_function.extent = 1.5
smoothing_function.smoothing_function = sCMOS.gaussian_filter_stack
smoothing_function.data_arg = "image"
```

```python
image_folder = (
    "/scratch2/jsb92/20250520_FRET/data/BSA_50pMDNAHJ/NotchFilter_785SP_10perc561_1"
)
image_files = H_F.file_search(image_folder, ".tif", "")
metadata_files = H_F.file_search(image_folder, "metadata", "")
```

```python
x_coord, y_coord, width, height = IO.metadata_reader_imageJ(metadata_files[0])
gain = gain_map[x_coord : x_coord + width, y_coord : y_coord + height]
offset = offset_map[x_coord : x_coord + width, y_coord : y_coord + height]
readnoise = read_noise[x_coord : x_coord + width, y_coord : y_coord + height]
rqe_cut = rqe[x_coord : x_coord + width, y_coord : y_coord + height]
```

```python
raw_data, first_frame, smoothed_frame, weights = IO.read_tiff_tophotoelectrons(
    image_files[0],
    frame=2,
    smoothing_function=smoothing_function,
    offset_map=offset,
    gain_map=gain,
    rqe=rqe_cut,
    read_noise=readnoise,
)
```
